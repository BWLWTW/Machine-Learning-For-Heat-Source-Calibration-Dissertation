# Block 1: Define Function

```python
import tensorflow as tf
tf.get_logger().setLevel('ERROR')
import keras
import pickle
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import time
def PRMSRE_global(y_true, y_pred):
    """
    Global PRMSRE over all elements in 2D arrays.
    """
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    mask = y_true != 0
    rel_errors = np.where(mask, ((y_true - y_pred) / y_true) ** 2, np.nan) # at location mask is true, do the cal, if it's where mask if false, return nan, end up form an array giving the value of output in each corresponding location
    return np.sqrt(np.nanmean(rel_errors))*100 # mean the non-nan ppl and sqrt them and *100
def AUTO_USE(working_folder,INPUT):
    ANN_loc=working_folder+r"\saved_model"
    SX_loc=working_folder+r"\scaler_X.pkl"
    SY_loc=working_folder+r"\scaler_Y.pkl"
    input_list=INPUT
    model=keras.models.load_model(ANN_loc)
    SX=pickle.load(open(SX_loc,'rb'))
    SY=pickle.load(open(SY_loc,'rb'))
    input_Scale=SX.transform(np.array([input_list]))
    answer_Scale=model.predict(input_Scale)
    answer=SY.inverse_transform(answer_Scale)
    return answer
def AUTO_CALIBRATION(working_folder_3,working_folder_2,ral,rll,rvl,IN,OUTPUT):
    test_r_record=[]
    test_loss_record=[]
    c=0
    start=time.time()
    for ra in ral:
        for rl in rll:
            for rv in rvl:
                print(c+1,'/',len(ral)*len(rll)*len(rvl),'Iterations:')
                r=[ra,rl,rv]
                test_r_record.append(r)
                answer=[float(AUTO_USE(working_folder_3,[IN[0],IN[1],ra,rl,rv])[0,0]),float(AUTO_USE(working_folder_2,[IN[0],IN[1],ra,rl,rv]))]
                print('current testing value:', r)
                Current_PRMSRE=PRMSRE_global(OUTPUT, answer)
                print('Current_PRMSRE:', Current_PRMSRE)
                test_loss_record.append(Current_PRMSRE)
                c+=1
    end=time.time()
    print('result:')
    i=test_loss_record.index(min(test_loss_record)) 
    print('using radii:',test_r_record[i])
    print('Corresponding PRMSRE:',test_loss_record[i])
    print('used time: ', end-start,' s.')
```
